```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("superstore.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
sns.countplot(x='Direction', data=df)

plt.title("Market Direction")

plt.show()
```

```python
year_volume = df.groupby('Year')['Volume'].mean()

year_volume.plot(kind='bar')

plt.title("Average Volume by Year")

plt.ylabel("Volume")

plt.show()
```

```python
plt.hist(df['Today'], bins=30)

plt.title("Today's Market Returns Distribution")

plt.xlabel("Today's Return")

plt.ylabel("Frequency")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=['number'])

sns.heatmap(numeric_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.scatterplot(x='Lag1', y='Today', data=df)

plt.title("Lag1 vs Today's Return")

plt.show()
```

# Insights

1. Market trading volume changed across different years.

2. Daily market returns showed both positive and negative fluctuations.

3. Some lag variables showed weak correlations with current market returns.

4. The dataset demonstrated trends useful for financial market analysis.

5. Correlation analysis helped identify relationships between market variables.
